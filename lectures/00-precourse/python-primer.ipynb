### What is a notebook?

So working with data needs a different kind of tool: one that lets you load a table once and then keep trying things on it, without loading the file again.

That tool is a **notebook**, and it is the thing almost everyone who works with data lives in.

Everything you create stays in the notebook's memory, so you rarely need to rerun the cells above. That memory is cleared when you close the notebook, of course.

### Variable types

## 4. Dictionaries

A **dictionary** (`dict`) stores labeled values: `key -> value`. Think "one row about one thing":

````python
product = {
    'name': 'Office chair', # (key='name') -> (value='Office chair')
    'price': 8990,
    'stock': 14,
}

# read a value by its key
print('Key = price, value =', product['price'])
product['category'] = 'Furniture'        # add a new key
product['stock'] = product['stock'] - 1  # update a value (one chair sold)

print(product) # print the whole dict
print(f"Stock value: {product['price'] * product['stock']} RUB")
````

## 5. Conditions

Let's start with a simple example: decide whether the company is profitable.

````python
margin_pct = 12.5 # pct

if margin_pct < 0:
    verdict = 'losing money'
elif margin_pct < 10:
    verdict = 'thin margin'
else:
    verdict = 'healthy'

# combining " and '
print(f"The company's margin is {margin_pct}% -> {verdict}")

# Comparisons: ==  !=  <  <=  >  >=   Combine with: and, or, not
print(margin_pct > 10 and margin_pct < 20)
````

## 6. Loops

A `for` loop repeats an action for each element in a sequence.

Python has a second kind of loop, `while`, which repeats as long as some condition stays true instead of walking through a sequence. For almost everything you will do, `for` is enough.

The two patterns below, **accumulation** and **filter-and-count**, cover a surprisingly large share of the data analysis code you will ever write.

````python
week = [1200, 950, 1430, 1100, 880, 1620, 1010]

# Pattern 1: ACCUMULATE — start from zero, add things up
total = 0
for day in week:
    total = total + day # or you can write total += day
print(f'Total: {total}')
````

````python
orders_week = [
    {'order_id': 1, 'city': 'Moscow', 'amount': 1500},
    {'order_id': 2, 'city': 'Kazan',  'amount': 700},
    {'order_id': 3, 'city': 'Moscow', 'amount': 2900},
    {'order_id': 4, 'city': 'Kazan',  'amount': 450},
    {'order_id': 5, 'city': 'Moscow', 'amount': 1100},
]

# Pattern 2: FILTER AND COUNT — loop + if + accumulate
moscow_revenue = 0
moscow_orders = 0
for order in orders_week:
    if order['city'] == 'Moscow':
        moscow_revenue += order['amount']   # += is shorthand for x = x + ...
        moscow_orders += 1

print(f'Moscow: {moscow_orders} orders, {moscow_revenue} RUB')
````

Two built-in functions make loops easier:
- `range(start, end, step)`, or just `range(n)` — when you need to repeat something `n` times
- `enumerate(sequence)` — when you also need the position of the current element

````python
# range(n) gives 0, 1, ..., n-1 — when you just need to repeat n times
for i in range(3):
    print(f'Day {i}')
````

````python
# Only odd days
for i in range(1, 8, 2):
    print(f'Day {i}')
````

````python
# with index
for i, order in enumerate(['order1', 'order2', 'order3']):
    print(f'Order={order}, index={i}')
````

### Walking a dictionary

A `for` loop walks a dictionary too. A dictionary holds pairs, so the loop takes two names at once: `.items()` gives you the key and the value together.

```python
for key, value in some_dict.items():
```

The other thing you keep asking a dictionary is whether a key is there at all. `key in some_dict` is `True` when the dictionary has that key, `key not in some_dict` is `True` when it does not. Asking first is what stops you from reading a key nobody ever created.

````python
# a toy dict: three products, small enough to check by eye
stock_by_product = {'chairs': 3, 'desks': 5, 'lamps': 2}

# .items() gives the key and the value together, one pair per turn
for product_name, quantity in stock_by_product.items():
    print(f'{product_name}: {quantity} in stock')

# 'in' asks whether a key exists at all
print('desks' in stock_by_product)
print('sofas' in stock_by_product)

if 'sofas' not in stock_by_product:
    print('No sofas in the dict yet')
````

## 7. Functions

A **function** wraps a computation so you can reuse it: inputs in parentheses, result via `return`.

We have already used several built-in Python functions, such as `max()`, `type()`, and `print()`. We have also used methods that belong to particular types such as `'Steve'.upper()`.

You can also write your own. If you find yourself copy-pasting a calculation, that is the signal to wrap it in a function:

````python
def profit(revenue, cost):
    """Profit in RUB. (This description is called a docstring.)"""
    return revenue - cost

def profit_margin(revenue, cost):
    """Profit as % of revenue."""
    return (revenue - cost) / revenue * 100

print(profit(34860, 17430))
print(f'{profit_margin(34860, 17430):.1f}%')

# Now it works for any numbers — including inside a loop:
for rev, cost in [[1000, 900], [5000, 2000], [300, 350]]:
    print(rev, cost, f'-> margin {profit_margin(rev, cost):.1f}%')
````

### Reading a CSV file with the built-in library

Python has many libraries.

A **library** is a collection of code written by someone else that you can use in your own programs. For example, you may already be familiar with `pandas`, a library that provides tools for working with tables.

Most libraries need to be installed before you can use them. However, Python also includes a **standard library**: a collection of modules that come with Python itself.

For example:

- `sys` — settings of Python itself
- `os` and `pathlib` — working with files, folders and paths
- `csv` — reading and writing CSV data


````python
import csv   # 'import' loads a library; csv ships with Python

total = 0
with open('sales.csv') as f:
    # each row arrives as a dict, the same shape as the order dicts above
    for row in csv.DictReader(f):
        print(row)
        total += float(row['amount'])   # values come as TEXT -> convert before math

print(f'Total: {total} RUB')
````

### Reading JSON

Think of JSON as a dictionary written out as text. The two have a lot in common.

````python
import json

# JSON is how systems talk to each other — e.g. an online store's API response:
api_response = '''{
  "order_id": 10023,
  "customer": {"name": "Anna", "city": "Moscow"},
  "items": [
    {"sku": "A-11", "qty": 2, "price": 990},
    {"sku": "B-07", "qty": 1, "price": 2450}
  ]
}'''

order = json.loads(api_response)          # text -> dicts and lists
print(order['customer']['city'])          # walk into nested structure step by step

order_total = 0
for item in order['items']:
    order_total += item['qty'] * item['price']
print(f'Order total: {order_total} RUB')
````

````python
# Solution — Exercise 5
def profit_margin_safe(revenue, cost):
    """Profit as % of revenue; None when revenue is 0."""
    if revenue == 0:
        return None
    return (revenue - cost) / revenue * 100

print(profit_margin_safe(1000, 600))
print(profit_margin_safe(0, 500))
````

````python
# Solution — Final challenge
def revenue_by_city(orders):
    """Revenue per city over delivered orders only."""
    revenue = {}
    for order in orders:
        if order['status'] == 'delivered':
            city = order['city']
            if city not in revenue:
                revenue[city] = 0
            revenue[city] += order['amount']
    return revenue

orders_with_status = [
    {'order_id': 1, 'city': 'Moscow', 'amount': 1500, 'status': 'delivered'},
    {'order_id': 2, 'city': 'Kazan',  'amount': 700,  'status': 'delivered'},
    {'order_id': 3, 'city': 'Moscow', 'amount': 2900, 'status': 'cancelled'},
    {'order_id': 4, 'city': 'Sochi',  'amount': 1200, 'status': 'delivered'},
    {'order_id': 5, 'city': 'Moscow', 'amount': 1100, 'status': 'delivered'},
    {'order_id': 6, 'city': 'Kazan',  'amount': 450,  'status': 'cancelled'},
    {'order_id': 7, 'city': 'Sochi',  'amount': 2100, 'status': 'delivered'},
]

result = revenue_by_city(orders_with_status)
for city, amount in result.items():
    print(f'{city}: {amount} RUB')

best_city = None
best_amount = 0
for city, amount in result.items():
    if amount > best_amount:
        best_city = city
        best_amount = amount

print(f'Best city: {best_city}')
````
